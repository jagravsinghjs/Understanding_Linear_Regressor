```python
import pandas as pd
import numpy as np
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../data/house_prices.csv')
X = df.drop('price', axis=1).values
y = df['price'].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=64)
```

```python
mean = np.mean(X_train, axis=0)
std  = np.std(X_train, axis=0)
std[std == 0] = 1
X_train_scaled = (X_train - mean) / std
X_test_scaled  = (X_test  - mean) / std
```

```python
X_train_b    = np.hstack((np.ones((X_train.shape[0], 1)), X_train))
X_test_b     = np.hstack((np.ones((X_test.shape[0],  1)), X_test))
```

```python
X_T     = np.transpose(X_train_b)
XtX     = np.dot(X_T, X_train_b)
XtX_inv = np.linalg.pinv(XtX)
XtY     = np.dot(X_T, y_train)
theta   = np.dot(XtX_inv, XtY)
```

```python
y_train_pred_n = np.dot(X_train_b, theta)
y_test_pred_n  = np.dot(X_test_b,  theta)
rmse_train_n   = root_mean_squared_error(y_train, y_train_pred_n)
rmse_test_n    = root_mean_squared_error(y_test,  y_test_pred_n)
print('Normal Equation — Train RMSE:', rmse_train_n)
print('Normal Equation — Test  RMSE:', rmse_test_n)
```

```python
import pickle

results_normal = {
    "rmse_train_n": rmse_train_n,
    "rmse_test_n": rmse_test_n,
    "theta": theta
}

with open("normal_eq_results.pkl", "wb") as f:
    pickle.dump(results_normal, f)
```
